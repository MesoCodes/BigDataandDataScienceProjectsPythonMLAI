```python
#Part 1
import matplotlib.pyplot as plt
import numpy as np
```

```python
labels = ['A', 'B', 'C', 'D', 'E']

Jan = np.array([20, 35, 30, 35, 27])
Feb = np.array([25, 40, 27, 10, 15])
Mar = np.array([15, 32, 30, 30, 10])
Apr = np.array([30, 12, 34, 20, 25])

plt.bar(labels, Jan, color='r', width=0.35)
plt.bar(labels, Feb, bottom=Jan, color='g', width=0.35)
plt.bar(labels, Mar, bottom=Jan+Feb, color='b', width=0.35)
plt.bar(labels, Apr, bottom=Jan+Feb+Mar, color='c', width=0.35)
plt.xlabel("Dealership")
plt.ylabel("Number of sold cars")
plt.legend(["January", "February", "March", "April"])
plt.title("Number of cars sold by dealers A, B, C, D, E")
plt.show()
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np
np.random.seed(42)

zdata_group_1 = 15 * np.random.random(100)
xdata_group_1 = np.sin(zdata_group_1) + 0.1 * np.random.randn(100)
ydata_group_1 = np.cos(zdata_group_1) + 0.1 * np.random.randn(100)

xdata_group_2 = [4, 9, 9, 5, 10, 8]
ydata_group_2 = [5, 5, 2, 10, 5, 3]
zdata_group_2 = [10, 3, 5, 8, 9, 6]

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(xdata_group_1, ydata_group_1, zdata_group_1, c='green', marker='o', label='Group 1')
ax.scatter(xdata_group_2, ydata_group_2, zdata_group_2, c='red', marker='+', label='Group 2')

ax.set_xlabel('x-axis')
ax.set_ylabel('y-axis')
ax.set_zlabel('z-axis')
ax.set_title('3D Scatter Plot Graph')
ax.legend()
plt.show()
```

```python
#Part 2
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np
np.random.seed(42)

zdata_group_1 = 15 * np.random.random(100)
xdata_group_1 = np.sin(zdata_group_1) + 0.1 * np.random.randn(100)
ydata_group_1 = np.cos(zdata_group_1) + 0.1 * np.random.randn(100)

zdata_group_2 = [10, 3, 5, 8, 9, 6]
xdata_group_2 = [4, 9, 9, 5, 10, 8]
ydata_group_2 = [5, 5, 2, 10, 5, 3]

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.scatter(xdata_group_1, ydata_group_1, zdata_group_1, c='blue', marker='o', s=50, alpha=0.7)
ax.scatter(xdata_group_2, ydata_group_2, zdata_group_2, c='red', marker='t', s=50, alpha=0.7)
ax.set_xlabel('x Axis')
ax.set_ylabel('y Axis')
ax.set_zlabel('z Axis')
ax.set_title('3D Scatter Plot Graph')

plt.show()
```
