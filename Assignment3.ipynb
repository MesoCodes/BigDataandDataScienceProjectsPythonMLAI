```python
import pandas as pd
Titanic = pd.DataFrame([["Children", "First", 6, 0],["Children", "Second", 24, 0],["Children", "Third", 27, 52],["Men", "First", 57, 118],["Men", "Second", 14, 154],["Men", "Third", 75, 387],["Men", "Crew", 192, 693],["Women", "First", 140, 4],["Women", "Second", 80, 13],["Women", "Third", 76, 89],["Women", "Crew", 20, 3]],
                        columns=["sex", "class", "survived", "died"])
print(Titanic)
```

```python
Titanic = Titanic.drop(Titanic[Titanic["class"] == "Crew"].index)
print(Titanic)
```

```python
Titanic["total no. people"] = Titanic["survived"] + Titanic["died"]
print(Titanic)
```

```python
Titanic = Titanic.drop("total no. people", axis=1)
print(Titanic)
```

```python
Titanic = Titanic[(Titanic["survived"] + Titanic["died"])*0.8 < Titanic["survived"]]
print(Titanic)
```
