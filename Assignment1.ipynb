```python
# Part 2
class Triangle(object):
    def __init__(self, base, height):
        self.base = base
        self.height = height
    def getBase(self):
        return self.base
    def setBase(self, b):
        self.base = b
    def getHeight(self):
        return self.height
    def setHeight(self,h):
        self.height = h
    def Area(self):
        return 0.5 * self.base * self.height
trianglea = Triangle(3, 4)
print(trianglea.getBase())
print(trianglea.getHeight())
print(trianglea.Area())
```
