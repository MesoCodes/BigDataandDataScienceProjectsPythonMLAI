```python
from joblib import Parallel, delayed

def primesquareorno(n):
    if n < 2:
        return None
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return None
    return n * n

if __name__ == "__main__":
    numbers = range(2, 100001) 
    num_cores = 4

results = Parallel(n_jobs=num_cores)(delayed(primesquareorno)(num) for num in numbers)

primesquare = [square for square in results if square is not None]

for square in primesquare:
        print(square)
```
